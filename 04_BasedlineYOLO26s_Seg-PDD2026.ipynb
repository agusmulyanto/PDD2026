```python
# ============================================================
# YOLO26 SEGMENTATION TRAINING (DEFAULT ULTRALYTICS)
# ============================================================

import gc
import os
from datetime import datetime

import matplotlib.pyplot as plt
from PIL import Image
from ultralytics import YOLO

# ============================================================
# CLEAR MEMORY
# ============================================================

gc.collect()

# ============================================================
# CONFIGURATION
# ============================================================

DATA_YAML = "FinalDataset_1024/data.yaml"
MODEL = "yolo26s-seg.pt"

RUN_NAME = f"YOLO26_SEG_{datetime.now().strftime('%Y%m%d_%H%M%S')}"

# ============================================================
# LOAD MODEL
# ============================================================

print("=" * 60)
print("Loading Model...")
print("=" * 60)

model = YOLO(MODEL)

# ============================================================
# TRAIN MODEL (DEFAULT ULTRALYTICS)
# ============================================================

print("Start Training...")
print("=" * 60)

results = model.train(
    data=DATA_YAML,
    epochs=400,
    imgsz=640,
    batch=8,
    workers=4,
    cache=False,
    device=0,
    amp=True
)


# ============================================================
# VALIDATION
# ============================================================

print("\n" + "=" * 60)
print("Validation Results")
print("=" * 60)

metrics = model.val()

print(metrics)

# ============================================================
# TEST PREDICTION
# ============================================================

print("\n" + "=" * 60)
print("Running Prediction on Test Images")
print("=" * 60)

model.predict(
    source="FinalDataset_1024/images/test",
    imgsz=640,
    conf=0.25,
    save=True,
    project="runs_yolo26_seg",
    name=f"{RUN_NAME}_predict"
)

# ============================================================
# VISUALIZATION
# ============================================================

save_dir = results.save_dir

print("\nTraining Results Saved To:")
print(save_dir)

plots = [
    "results.png",
    "confusion_matrix.png",
    "PR_curve.png",
    "P_curve.png",
    "R_curve.png",
    "F1_curve.png",
]

for plot in plots:

    path = os.path.join(save_dir, plot)

    if os.path.exists(path):

        plt.figure(figsize=(10,8))
        plt.imshow(Image.open(path))
        plt.title(plot.replace(".png",""))
        plt.axis("off")
        plt.show()

print("=" * 60)
print("Training Finished Successfully")
print("=" * 60)
```
